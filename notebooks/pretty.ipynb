```python
from cdxcore.pretty import PrettyObject
import pickle

po = PrettyObject()

po.x = 3
po.f = lambda self, y : self.x*y
po['g'] = lambda y : y

print( po.f(y=2), po.g(7) )
```

```python
class A( PrettyObject ):
    def f(self,y):
        return self.x*y
    @staticmethod
    def g(z):
        return z

a = A(x=3)
print( a.f(2) )
print( a.g(7) )
```

```python
class AO( PrettyObject ):
    def __init__(self, x):
        self.x = x
        print(self.__init__.__qualname__)
po = AO(x=3)

sdir = SubDir("!/test", create_directory=True )
with open( sdir.full_file_name("po"), "wb") as f:
    pickle.dump(po, f)
    
sdir = SubDir("!/test", create_directory=True )
with open( sdir.full_file_name("po"), "rb") as f:
    po2 = pickle.load(f)

```

```python
from cdxcore.pretty import PrettyObject
pdct = PrettyObject()
pdct.x = 1
pdct['y'] = 2
print( pdct['x'], pdct.y ) # -> prints 1 2

print(type(pdct.at_pos))
print(type(pdct.at_pos.keys))
print(type(pdct.at_pos.values))
print(type(pdct.at_pos.items))

print(pdct.at_pos[1])          # -> prints 2
print(pdct.at_pos.keys[1])     # -> prints "y"
print(list(pdct.at_pos.items)) # -> prints "y"

assert pdct.at_pos[0] == 1
assert pdct.at_pos[1] == 2
assert pdct.at_pos.keys[0] == "x"
assert pdct.at_pos.keys[1] == "y"
assert pdct.at_pos.values[0] == 1,pdct.at_pos.values[0]
assert pdct.at_pos.values[1] == 2,pdct.at_pos.values[1]
assert pdct.at_pos.items[0] == ("x",1)
assert pdct.at_pos.items[1] == ("y",2)
```

```python
from cdxcore.pretty import PrettyObject
from dataclasses import dataclass
@dataclass 
class A:
    data : PrettyObject = PrettyObject(x=2).as_field()

a = A() 
print(a.data.x)  # -> "2"
a = A(data=PrettyObject(x=4)) 
print(a.data.x)  # -> "4"
```
